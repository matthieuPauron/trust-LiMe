# Error_TwoWayNoNS — two-way velocity without Navier-Stokes fails cleanly

`MagneticCoupled` · ASN 28 negative (diagnostic) test case (TRUST-LiMe).


## Introduction

Negative case: the `vitesse_hydraulique` keyword takes the transporting velocity
from the `Navier_Stokes_standard` equation of the same problem.  A data file
carrying the keyword but no `Navier_Stokes_standard` block must be rejected.

## Contracts checked

| # | Check |
|---|:--|
| 1 | non-zero exit + readable message |
| 2 | the diagnostic names `vitesse_hydraulique` and `Navier_Stokes_standard` |


```python
from trustutils import run

run.TRUST_parameters("1.9.8")
run.reset()

```

## Run

Run **sequentially** (no parallel launch) so the mesh is built
inline and the missing-equation diagnostic is reached.


```python
import subprocess, tempfile, shutil, os

CASE = "error_two_way_no_ns"

tmp = tempfile.mkdtemp(prefix=f"lime_{CASE}_")
shutil.copy(os.path.join("src", f"{CASE}.data"), tmp)
try:
    p = subprocess.run(["trust", CASE], cwd=tmp,
                       capture_output=True, text=True, timeout=300)
    rc, out = p.returncode, (p.stdout + p.stderr)
finally:
    shutil.rmtree(tmp, ignore_errors=True)

assert rc != 0, f"{CASE}: expected a non-zero exit, got {rc}"
assert out.strip(), f"{CASE}: no diagnostic message emitted"
assert "vitesse_hydraulique" in out and "Navier_Stokes_standard" in out, \
    f"{CASE}: the diagnostic must name vitesse_hydraulique/Navier_Stokes_standard"
tail = [l for l in out.splitlines() if l.strip()][-1:] or [""]
print(f"{CASE}: rc={rc}  |  {tail[0][:90]}")
print("OK: two-way velocity without Navier-Stokes rejected with a readable message.")

```

## Result


```python
# Negative (diagnostic) case: run directly with `trust`.
print("Negative case complete: missing Navier-Stokes equation rejected with a non-zero exit.")
print(f"case: {CASE}  rc={rc}")

```

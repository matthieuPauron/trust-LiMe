# Error_TwoWayBothVelocities — ambiguous transport velocity fails cleanly

`MagneticCoupled` · ASN 28 negative (diagnostic) test case (TRUST-LiMe).


## Introduction

Negative case: the transport-velocity source must be unambiguous.  A data
file declaring both `vitesse_magnetique` (data field) and
`vitesse_hydraulique` (two-way flag) in the magnetic equation block must be
rejected while the block is read.

## Contracts checked

| # | Check |
|---|:--|
| 1 | non-zero exit + readable message |
| 2 | the diagnostic reports the mutual exclusion |


```python
from trustutils import run

run.TRUST_parameters("1.9.8")
run.reset()

```

## Run

Run **sequentially** (no parallel launch) so the mesh is built
inline and the ambiguous-keyword diagnostic is reached.


```python
import subprocess, tempfile, shutil, os

CASE = "error_two_way_both_velocities"

tmp = tempfile.mkdtemp(prefix=f"lime_{CASE}_")
shutil.copy(os.path.join("src", f"{CASE}.data"), tmp)
try:
    p = subprocess.run(["trust", CASE], cwd=tmp,
                       capture_output=True, text=True, timeout=300)
    rc, out = p.returncode, (p.stdout + p.stderr)
finally:
    shutil.rmtree(tmp, ignore_errors=True)

assert rc != 0, f"{CASE}: expected a non-zero exit, got {rc}"
assert out.strip(), f"{CASE}: no diagnostic message emitted"
assert "mutually exclusive" in out, \
    f"{CASE}: the diagnostic must report the mutual exclusion"
tail = [l for l in out.splitlines() if l.strip()][-1:] or [""]
print(f"{CASE}: rc={rc}  |  {tail[0][:90]}")
print("OK: ambiguous transport velocity rejected with a readable message.")

```

## Result


```python
# Negative (diagnostic) case: run directly with `trust`.
print("Negative case complete: ambiguous velocity keywords rejected with a non-zero exit.")
print(f"case: {CASE}  rc={rc}")

```

## Introduction

Negative (diagnostic) case of the **standalone magnetic (induction) equation** added by TRUST-LiMe.

## What it validates

This notebook runs **one and only one test case**: `error_bad_nbcomposantes`, where an invalid component count is rejected. A malformed dataset must terminate with a non-zero exit code and a readable diagnostic, rather than crashing silently.

The case is run **directly** with the `trust` command — not through `runCases()`, which would abort the notebook on the non-zero exit.

## Contracts checked

| # | Modelling decision | Check |
|---|---|:--:|
| 11 | malformed input fails cleanly | non-zero exit + readable message |


```python
from trustutils import run

run.TRUST_parameters("1.9.8")
run.reset()
```

## Run


```python
import subprocess, tempfile, shutil, os

CASE = "error_bad_nbcomposantes"

tmp = tempfile.mkdtemp(prefix=f"lime_{CASE}_")
shutil.copy(os.path.join("src", f"{CASE}.data"), tmp)
try:
    p = subprocess.run(["trust", CASE, "2"], cwd=tmp,
                       capture_output=True, text=True, timeout=300)
    rc, out = p.returncode, (p.stdout + p.stderr)
finally:
    shutil.rmtree(tmp, ignore_errors=True)

# a clean TRUST failure prints a diagnostic and exits non-zero
assert rc != 0, f"{CASE}: expected a non-zero exit, got {rc}"
assert out.strip(), f"{CASE}: no diagnostic message emitted"
tail = [l for l in out.splitlines() if l.strip()][-1:] or [""]
print(f"{CASE}: rc={rc}  |  {tail[0][:90]}")
print("OK: malformed input rejected with a readable message.")
```

## Result


```python
# Negative (diagnostic) case: run directly with `trust`, so there is no
# trustutils suite and no performance table. Report the outcome instead.
print("Negative case complete: malformed input rejected with a non-zero exit.")
print(f"case: {CASE}  rc={rc}")
```

# SmoothDiskMHD: 3D coupled MHD in a box

Laminar proxy (Rm ~ 1) of the VKS experiment: Cartesian box with two
counter-rotating smooth disks (modelized propellers, `Paroi_defilante`),
insulating magnetic walls, two-way coupling (`force_lorentz` +
`vitesse_hydraulique`). No blade geometry. Runs on 2 ranks with
partitioning. Smoke case only (no analytic assert).

```python
from trustutils import run

run.reset()
case = run.addCaseFromTemplate("smooth_disk_mhd.data", "coarse", dic={},
                              nbProcs=2, excluNR=True)
case.partition()
run.printCases()
run.runCases()
```

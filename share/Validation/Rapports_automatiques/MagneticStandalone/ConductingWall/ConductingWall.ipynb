# ConductingWall — hydraulic k-ε feeds the magnetic eddy diffusivity

`MagneticStandalone` · Phase 4 validation (TRUST-LiMe).


## Introduction

Standalone magnetic (induction) equation with a turbulent hydraulic
momentum equation: the magnetic eddy diffusivity ηₜ is grabbed from the
hydraulic k-ε model (no explicit k/eps fields) and folded into η_eff.

## What it validates

- the `Pb_MHD_Turbulent` problem class (NS turbulent + magnetic);
- the `vitesse_hydraulique` binding on VEF (triangular mesh);
- the grab message `hydraulic turbulence model detected, using its
  k/epsilon for the magnetic eddy diffusivity (eta_t in [min, max])`;
- the non-uniform fold `turbulent magnetic diffusivity folded (non-uniform)
  into eta_eff = eta + eta_t`.
